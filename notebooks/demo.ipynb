# Stage 1: Rule-Based Parser Demonstration

This notebook shows the symbolic Stage 1 implementation of the shoe-world system. The Commands are interpreted using the `RuleBasedParser`, grounded against the objects and relations in the current world, and executed through the shared dialogue and action pipeline.

A reproducible world is generated using a fixed random seed (205). All commands are executed sequentially in the same world, so successful actions update the world state and can affect later commands.

The examples demonstrate:
* object references based on attributes, locations, object ids, and relations
* all possible world actions are shown
* ambiguity detection and unsuccessful reference resolution
* precondition checking, action execution, and meaningful system messages

This demonstration uses only the symbolic rule-based language-understanding component.


```python
import sys
import importlib
from pathlib import Path
from typing import Dict, Any, List, Optional, Tuple

PROJECT_ROOT = Path.cwd().resolve()
if not (PROJECT_ROOT / "src").is_dir():
    PROJECT_ROOT = PROJECT_ROOT.parent
sys.path.insert(0, str(PROJECT_ROOT))

from src import dialogue, parsers, world

EVALUATION_SEED = 205
```

```python
DEMO_COMMANDS = [
    "Pick up the shoe next to the wet sneaker",
    "Put down the shoe on the middle shelf",
    "move the white shoe from the bottom shelf into the drying area",
    "dry the brown rubber shoe",
    "please clean the very dirty shoe with a brush that has at least 25% fullness",
    "impregnate the sandal that has the intact sole using the spray",
    "repair the sole of the red sandal with the sole hammer",
    "go on a long walk with boot_1 at the construction site in rainy weather",
    "get a new needle from the infinite tool box",
    "repair the white shoe with the sole glue",
    "pick up the white sneaker",
    "pick up the white boot",
    "put down the black boot on the floor",
    "go on a walk with the white sneaker",
    "use the thing"
]



#------------------------------------------------------------------
# Execute rule-based example commands sequentially in a fixed world
#------------------------------------------------------------------
    
print("Rule-Based Parsing Demonstration \n" + "#" * 100 + "\n")

# Use a fixed seed so the demonstration is reproducible
demo_world = world.World.create_random(seed=EVALUATION_SEED)
parser = parsers.RuleBasedParser()
    
print("#" * 100 + "\nINITIAL WORLD\n" + "#" * 100 + "\n")
print(demo_world.describe() + "\n\n")

# Reuse the same world so successful commands affect later interactions
for number, command in enumerate(DEMO_COMMANDS, start=1):
    print("#" * 100 + f"\nCOMMAND {number}: {command}\n" + "#" * 100)
    try:
        result = dialogue.interpret_and_act(
            world=demo_world,
            utterance=command,
            parser=parser
        )
        print(dialogue.format_dialogue_output(result) + "\n")
    except Exception as error:
        print(f"ERROR: {type(error).__name__}: {error}")
```
